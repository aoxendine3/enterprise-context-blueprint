```python
# Cell 2: Feature Engineering & Athletic Explosiveness Index (AEI)
merged = pd.merge(players_df, combine_df, on='nfl_id', how='inner')
merged = pd.merge(merged, success_df, on='nfl_id', how='left')

numeric_cols = ['forty', 'ten_yd_split', 'vertical', 'broad_jump', 'three_cone', 'short_shuttle', 'ngs_athleticism_score']
for col in numeric_cols:
    merged[col] = pd.to_numeric(merged[col], errors='coerce')

# Compute AEI Score
merged['aei_score'] = (merged['vertical'] / (merged['forty'] + 1e-5)) + (merged['broad_jump'] / (merged['ten_yd_split'] * 10 + 1e-5))
merged['aei_score'] = merged['aei_score'].fillna(merged['aei_score'].median())
merged['position_rank'] = merged.groupby('nfl_position')['aei_score'].rank(ascending=False, method='min')

submission = merged[['nfl_id', 'display_name', 'nfl_position', 'aei_score', 'position_rank']].copy()
submission.to_csv('submission.csv', index=False)
print(f'Successfully exported submission.csv with {len(submission)} rows.')
print('Sample output predictions:')
print(submission.head(10))
```
