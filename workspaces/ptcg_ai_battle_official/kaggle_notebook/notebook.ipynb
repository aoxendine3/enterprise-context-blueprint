```python
# Cell 1: Extract Official Submission Package (2.1 MB Engine + Deck + API)
import base64
import os
import tarfile

SUBMISSION_B64 = ''''''

tar_bytes = base64.b64decode(SUBMISSION_B64)
with open('submission.tar.gz', 'wb') as f:
    f.write(tar_bytes)

print(f'Successfully wrote submission.tar.gz ({len(tar_bytes):,} bytes).')

# Extract for validation test run
os.makedirs('submission_ext', exist_ok=True)
with tarfile.open('submission.tar.gz', 'r:gz') as tar:
    tar.extractall('submission_ext')
print('Extracted submission contents:', os.listdir('submission_ext'))
```

```python
# Cell 2: Run Live Match Verification via Official C++ Engine
import sys
import os
import random

sys.path.insert(0, 'submission_ext')
from cg.game import battle_start, battle_select, battle_finish, visualize_data
from cg.api import to_observation_class, Observation

def read_deck():
    with open('submission_ext/deck.csv', 'r') as f:
        return [int(x.strip()) for x in f.readlines() if x.strip()][:60]

deck0 = read_deck()
deck1 = read_deck()

print(f'Starting official PTCG validation match with 60-card decks...')
obs_dict, start_data = battle_start(deck0, deck1)
step = 0
while obs_dict and 'winner' not in obs_dict:
    step += 1
    obs = to_observation_class(obs_dict)
    if obs.select is None:
        break
    options = obs.select.option
    choice = random.sample(list(range(len(options))), min(obs.select.maxCount, len(options)))
    obs_dict = battle_select(choice)

print(f'Validation match completed in {step} turns cleanly!')
viz = visualize_data()
print(f'Engine visualization log size: {len(viz):,} bytes.')
battle_finish()
```
