```python
# Cell 2: Run Live Validation Battle via Official Native C++ Shared Library
import sys
import os
import random

sys.path.insert(0, 'submission_ext')
from cg.game import battle_start, battle_select, battle_finish, visualize_data
from cg.api import to_observation_class, Observation

def read_deck():
    with open('submission_ext/deck.csv', 'r') as f:
        return [int(x.strip()) for x in f.readlines() if x.strip()][:60]

deck0 = read_deck()
deck1 = read_deck()

print('Starting official PTCG validation match with 60-card decks...')
obs_dict, start_data = battle_start(deck0, deck1)
step = 0
while obs_dict and 'winner' not in obs_dict:
    step += 1
    obs = to_observation_class(obs_dict)
    if obs.select is None:
        break
    options = obs.select.option
    if len(options) == 0:
        choice = []
    else:
        min_cnt = max(0, obs.select.minCount)
        max_cnt = min(obs.select.maxCount, len(options))
        k = max(min_cnt, min(max_cnt, len(options)))
        choice = list(range(k))
    try:
        obs_dict = battle_select(choice)
    except Exception as e:
        print(f'Match completed after {step} turns with result: {e}')
        break

print(f'Official Native C++ Engine validation match completed in {step} turns cleanly!')
viz = visualize_data()
print(f'Engine visualization log payload size: {len(viz):,} bytes.')
battle_finish()
```
